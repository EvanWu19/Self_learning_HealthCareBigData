# Postman / cURL Drills: FHIR‑Dev Environment Setup
This notebook summarizes the workflow for:
1. Creating a FHIR‑Dev environment in Postman
2. Importing the SMART R4 collection
3. Configuring OAuth2 authorization
4. Sending CRUD calls and inspecting responses
5. Exporting environment and collection, committing to Git
6. Mirroring requests with cURL and capturing shell history
7. Writing a reflection note

## Step 1: FHIR‑Dev Environment
- Open Postman > Manage Environments > Add New
- Define variables:
  - `base_url`: `http://localhost:8080/fhir`
  - (Optional) `client_id`, `client_secret`, `scope`, `redirect_uri`, `auth_token`
- Save and export as `postman/fhir_dev.postman_environment.json`

## Step 2: Import SMART on FHIR R4 Collection
- Download from:
  - GitHub: https://github.com/HL7/fhir-postman-collection (`smart-on-fhir-r4.collection.json`)
  - Or via Postman Explore: search “SMART on FHIR R4”
- Import into Postman (File > Import)

## Step 3: Configure Authorization
- In a request, set Authorization type to OAuth 2.0
- Grant Type: Authorization Code
- Auth URL: `{{base_url}}/oauth2/authorize`
- Token URL: `{{base_url}}/oauth2/token`
- Client ID/Secret, Scope, Callback: use environment variables
- Click **Get New Access Token** and **Use Token**

## Step 4: Send First CRUD Call
- Select FHIR‑Dev environment
- In **Create Patient** request, use URL: `{{base_url}}/Patient`
- Headers:
  - `Authorization: Bearer {{auth_token}}`
  - `Content-Type: application/fhir+json`
  - `Accept: application/fhir+json`
- Click **Send** and inspect **201 Created** response

## Step 5: Export & Commit
- Export environment & collection JSONs into your Git repo under `postman/`
- Run:
  ```bash
  git add postman/fhir_dev.postman_environment.json postman/smart-on-fhir-r4.collection.json
  git commit -m "Add Postman SMART R4 drills"
  ```
- Note the commit hash for deliverable

## Step 6: Mirror with cURL
- In Postman, click **Code** > **cURL** for each request
- Paste and run in shell:
  ```bash
  curl -X POST {{base_url}}/Patient ...
  ```
- Capture commands via `history`

## Step 7: Reflection Note
- Create `notes/2025-06-30_postman.md` summarizing:
  - Goals, Challenges, Learnings, Next Steps
- Example outline in conversation
