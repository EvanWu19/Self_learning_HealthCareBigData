# Postman cURL Demonstration

This notebook shows how the Postman CRUD calls can be mirrored using `curl`.

```
curl -X POST {{baseUrl}}/Patient -H 'Content-Type: application/fhir+json' -d @patient.json
```

```
curl -X GET {{baseUrl}}/Patient/123
```

```
curl -X PUT {{baseUrl}}/Patient/123 -H 'Content-Type: application/fhir+json' -d @updated_patient.json
```

```
curl -X PATCH {{baseUrl}}/Patient/123 -H 'Content-Type: application/json-patch+json' -d @patch.json
```

```
curl -X DELETE {{baseUrl}}/Patient/123
```
